```python
!pip install ultralytics -q

import os
from google.colab import files, output
from IPython.display import Audio, display
import numpy as np
from ultralytics import YOLO

# 1. ربط الـ Google Drive
if not os.path.exists('/content/drive'):
    from google.colab import drive
    drive.mount('/content/drive')

# 2. تحميل الموديل
MODEL_PATH = "/content/drive/MyDrive/ADAS_YOLOv26_Model/vision_yolo26_scratch_100epochs/weights/best.pt"
model = YOLO(MODEL_PATH)

# دالة لتوليد تنبيه صوتي (Beep) عبر بايثون
def play_alert_sound():
    samplerate = 8000
    t = np.linspace(0, 0.3, int(samplerate * 0.3))
    audio_data = np.sin(2 * np.pi * 880 * t)
    display(Audio(audio_data, rate=samplerate, autoplay=True))

# 3. زر تفاعلي لرفع فيديو الطريق من جهازك
print("📤 الرجاء اختيار ورفع فيديو (.mp4) من جهازك:")
uploaded_vids = files.upload()

for filename in uploaded_vids.keys():
    print(f"🎬 جاري معالجة الفيديو '{filename}' إطاراً بتلو الآخر مع تفعيل التنبيهات...")

    results = model.predict(
        source=filename,
        imgsz=1280,
        conf=0.25,
        save=True,
        device='cuda',
        project="/content/drive/MyDrive/ADAS_YOLOv26_Model",
        name="Inference_Video_Output",
        exist_ok=True,
        stream=True
    )

    frame_count = 0

    for r in results:
        frame_count += 1
        boxes = r.boxes
        if boxes is not None and len(boxes) > 0:
            for c in boxes.cls:
                class_name = model.names[int(c)]
                if class_name in ['no_entry', 'traffic_signal', 'stop_sign']:
                    if frame_count % 30 == 0:
                        # تم تصحيح سطر الطباعة هنا لإزالة الرمز المسبب للخطأ
                        print(f"ALERT: Detected {class_name}")
                        play_alert_sound()

    print(f"🎉 مبروك! تم الانتهاء من معالجة {frame_count} إطاراً وحفظ الفيديو في الدرايف:")
    print("📂 ADAS_YOLOv26_Model/Inference_Video_Output/")
```

```python
import os
import glob
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import drive

drive.mount('/content/drive')

print("Searching for 'results.csv' across Google Drive...")

# Search automatically for the experiment results file.
found_files = glob.glob(
    '/content/drive/MyDrive/**/results.csv',
    recursive=True
)

if not found_files:
    found_files = glob.glob('/content/drive/**/*.csv', recursive=True)
    found_files = [f for f in found_files if 'results' in f]

if found_files:
    results_csv_path = found_files[0]
    print(f"Results file found at:\n{results_csv_path}")

    df = pd.read_csv(results_csv_path)
    df.columns = df.columns.str.strip()

    output_dir = os.path.dirname(results_csv_path)

    # Configure a clean visualization style.
    plt.style.use(
        'seaborn-v0_8-whitegrid'
        if 'seaborn-v0_8-whitegrid' in plt.style.available
        else 'default'
    )

    # 1. Training and validation loss curves.
    plt.figure(figsize=(10, 6), dpi=300)

    plt.plot(
        df['epoch'],
        df['train/box_loss'],
        label='Train Box Loss',
        color='#1f77b4',
        linewidth=2
    )

    plt.plot(
        df['epoch'],
        df['train/cls_loss'],
        label='Train Class Loss',
        color='#ff7f0e',
        linewidth=2
    )

    plt.plot(
        df['epoch'],
        df['val/box_loss'],
        label='Validation Box Loss',
        color='#2ca02c',
        linestyle='--',
        linewidth=2
    )

    plt.plot(
        df['epoch'],
        df['val/cls_loss'],
        label='Validation Class Loss',
        color='#d62728',
        linestyle='--',
        linewidth=2
    )

    plt.title(
        'Training and Validation Loss Curves (100 Epochs)',
        fontsize=14,
        fontweight='bold',
        pad=15
    )
    plt.xlabel('Epochs', fontsize=12)
    plt.ylabel('Loss Value', fontsize=12)
    plt.legend(fontsize=10, loc='upper right')
    plt.tight_layout()

    loss_plot_path = os.path.join(
        output_dir,
        "loss_curves_academic.png"
    )

    plt.savefig(loss_plot_path)
    plt.show()

    # 2. mAP performance curves.
    plt.figure(figsize=(10, 6), dpi=300)

    map50_col = next(
        (
            c for c in df.columns
            if 'mAP50(B)' in c or 'mAP50' in c
        ),
        None
    )

    map50_95_col = next(
        (
            c for c in df.columns
            if 'mAP50-95(B)' in c or 'mAP50-95' in c
        ),
        None
    )

    if map50_col and map50_95_col:
        plt.plot(
            df['epoch'],
            df[map50_col],
            label='mAP50',
            color='#9467bd',
            linewidth=2.5
        )

        plt.plot(
            df['epoch'],
            df[map50_95_col],
            label='mAP50-95',
            color='#8c564b',
            linewidth=2.5
        )

    plt.title(
        'Model Performance Progression (mAP)',
        fontsize=14,
        fontweight='bold',
        pad=15
    )
    plt.xlabel('Epochs', fontsize=12)
    plt.ylabel('Metric Score', fontsize=12)
    plt.legend(fontsize=10, loc='lower right')
    plt.ylim(0, 1.05)
    plt.tight_layout()

    map_plot_path = os.path.join(
        output_dir,
        "map_curves_academic.png"
    )

    plt.savefig(map_plot_path)
    plt.show()

    # 3. Precision and recall curves.
    p_col = next(
        (
            c for c in df.columns
            if 'precision(B)' in c or 'precision' in c
        ),
        None
    )

    r_col = next(
        (
            c for c in df.columns
            if 'recall(B)' in c or 'recall' in c
        ),
        None
    )

    if p_col and r_col:
        plt.figure(figsize=(10, 6), dpi=300)

        plt.plot(
            df['epoch'],
            df[p_col],
            label='Precision',
            color='#2ca02c',
            linewidth=2.5
        )

        plt.plot(
            df['epoch'],
            df[r_col],
            label='Recall',
            color='#ff7f0e',
            linewidth=2.5
        )

        plt.title(
            'Precision and Recall Progression (100 Epochs)',
            fontsize=14,
            fontweight='bold',
            pad=15
        )
        plt.xlabel('Epochs', fontsize=12)
        plt.ylabel('Metric Score', fontsize=12)
        plt.legend(fontsize=10, loc='lower right')
        plt.ylim(0, 1.05)
        plt.tight_layout()

        pr_plot_path = os.path.join(
            output_dir,
            "precision_recall_academic.png"
        )

        plt.savefig(pr_plot_path)
        plt.show()

    print(
        f"\nThree performance plots were generated and saved successfully "
        f"in the experiment directory:\n{output_dir}"
    )

else:
    print("The 'results.csv' file was not found in Google Drive.")
```
